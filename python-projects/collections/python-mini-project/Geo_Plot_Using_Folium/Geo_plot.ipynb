1. Create Basic Webmap

````python
map = fo.Map()
map
````

2. Insert Child

````python
x = fo.FeatureGroup(name='My Map')
````

````python
x.add_child(fo.Marker(location=[27.1751,78.0421] , popup='hey', icon=fo.Icon(color='blue')))
````

````python
map.add_child(x)
````

3. Multiple Markers

````python
for lat,lon in ([34,53],[24,-50],[90,-68]):
    x.add_child(fo.Marker(location=[lat,lon] , 
                          popup='hey', icon=fo.Icon(color='red')))
````

````python
map.add_child(x)
````

## Plotting Volcano Locations

````python
map = fo.Map()
map
````

Read the ```volcano.csv``` file:

````python
volcano = pd.read_csv('volcano.csv')
````

Inspect the first 5 rows of the dataset:

````python
volcano.head()
````

**1. General View**

Create object variables:

````python
lat_vo = list(volcano['Latitude'])
lon_vo = list(volcano['Longitude'])
name_vo = list(volcano['Name'])
````

````python
vol = fo.FeatureGroup(name = 'My Map')
````

````python
for lat, lon, name in zip(lat_vo, lon_vo, name_vo):
    vol.add_child(fo.Marker(location=[lat,lon] , 
                          popup= name, icon=fo.Icon(color='red')))
````

Plot the map:

````python
map.add_child(vol)
````

**2. Polygen View**

````python
vol.add_child(fo.GeoJson(data=(open('world.json','r',encoding='utf-8-sig').read())))
````

````python
map.add_child(vol)
````

## US Cities Population

Read the ```us+cities+pop.csv``` file:

````python
popu = pd.read_csv('us+cities+pop.csv')
````

````python
popu.head()
````

Create object variables:

````python
lat_po = list(popu['lat'])
lon_po = list(popu['lon'])
name_po = list(popu['name'])
pop_po = list(popu['pop'])
````

````python
po = fo.FeatureGroup(name = 'My Map')
````

Define function for filtering cities by population number:

````python
def mar(popu):
    if(popu>35000):
        return 'red'
    elif(popu>10000 and popu<=35000):
        return 'blue'
    else:
        return 'green'
````

````python
for lat, lon, name, pop in zip(lat_po, lon_po, name_po, pop_po):
    po.add_child(fo.Marker(location=[lat,lon] , 
                          popup= name, icon=fo.Icon(color= mar(pop))))
````

Plot the map:

````python
map.add_child(po)
````
